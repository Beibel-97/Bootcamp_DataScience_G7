# PRUEBA DE USO DE RED NEURONAL CONVOLUCIONAL ENTRENADA

```python
import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# CARGAMOS EL MODELO

```python
red = keras.models.load_model('/content/cnn_neumonia.keras')
red.summary()
```

```python
PATH_IMAGEN_NEUMONIA = '/content/ryct.2020200034.fig2.jpeg'
```

```python
from keras.preprocessing import image
path_imagen_prueba = PATH_IMAGEN_NEUMONIA
imagen_prueba = image.load_img(path_imagen_prueba,target_size=(64,64))

test_image = image.img_to_array(imagen_prueba)
test_image = np.expand_dims(test_image,axis=0)

result = red.predict(test_image)
print(f'predicción : {result}')
```
