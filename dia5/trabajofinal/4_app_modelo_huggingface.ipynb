```python
from transformers import pipeline

pipe = pipeline("text-classification", model="cesarcodigo/modelo_sentimiento_peruano")
```

```python
texto = 'Esta peli esta buenaza'
respuesta = pipe(texto)
print(respuesta)
```

#CREAMOS APP CON GRADIO PARA EXPONER EL MODELO

```python
def analisis_sentimiento(texto):
  resultado = pipe(texto)[0]
  label = resultado['label']
  score = resultado['score']

  if label == 'positive':
    label = 'Positivo'
  else:
    label = 'Negativo'

  return f'Tú comentario  es {label} (score : {round(score, 4)})'
```

```python
print(analisis_sentimiento(texto))
```

```python
import gradio as gr

demo = gr.Interface(
    fn=analisis_sentimiento,
    inputs=gr.Textbox(label='coloca tu comentario'),
    outputs=gr.Textbox(label='Resultado')
)
demo.launch(
    share=True
)
```
