# 📘 DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

DBSCAN es un algoritmo de clustering basado en densidad.

A diferencia de KMeans:
- ❌ No necesita definir el número de clusters (K).
- ✅ Puede detectar ruido (outliers).
- ✅ Puede encontrar clusters con formas arbitrarias.

---

# 🔎 ¿Cómo funciona?

DBSCAN clasifica los puntos en tres tipos:

1️⃣ Punto núcleo (Core Point)
   Tiene al menos "min_samples" puntos dentro de su radio "eps".

2️⃣ Punto frontera (Border Point)
   Está dentro del radio de un punto núcleo.

3️⃣ Ruido (Noise)
   No pertenece a ningún cluster (etiqueta = -1).

---

# 🎯 Parámetros principales

- eps → radio de vecindad
- min_samples → número mínimo de puntos para formar un cluster
- metric → métrica de distancia

---

# 📌 Ventajas

✔ No necesita K  
✔ Detecta outliers  
✔ Encuentra clusters no esféricos  

---

# ⚠️ Desventajas

❌ Sensible a la escala  
❌ Difícil elegir eps  
❌ No funciona bien si densidades son muy distintas  


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

# 1️⃣ Cargar dataset

```python
iris = sns.load_dataset("iris")
X = iris.drop(columns=["species"])
```

# 2️⃣ Escalar (MUY IMPORTANTE)

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

# 3️⃣ Aplicar DBSCAN

```python
dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(X_scaled)
```

# 4️⃣ Visualizar

```python
plt.figure(figsize=(8,6))
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=labels,
    cmap="viridis",
    edgecolors="k",
    alpha=0.7
)
plt.xlabel("Característica 1 (escalada)")
plt.ylabel("Característica 2 (escalada)")
plt.title("Clusters con DBSCAN")
plt.show()
```

```python
print("Silhouette Score:", silhouette_score(X_scaled, labels))
```
