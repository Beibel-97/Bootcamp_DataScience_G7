## 1️⃣ ¿Qué significa “pendiente”?

La **pendiente** nos dice qué tan rápido cambia una variable respecto a otra.

Ejemplo:
- Si al aumentar `x` en 1, `y` aumenta en 2  
👉 la pendiente es **2**

En una recta:
$
y = mx + b
$

- `m` es la pendiente
- `b` es el punto donde la recta corta el eje `y`

---

## 2️⃣ ¿Qué pasa si la gráfica no es una recta?

Cuando la gráfica es una **curva**, la pendiente:
- No es constante
- Cambia en cada punto

Entonces surge la pregunta:

> ❓ ¿Cómo calculamos la pendiente en **un solo punto**?

Aquí aparece la idea de **límite**.

---

## 3️⃣ Idea intuitiva de límite

Tomamos dos puntos muy cercanos:

- `h` es un número muy pequeño
- Mientras `h` se acerca a 0, la pendiente se vuelve más precisa

👉 El **límite** estudia qué pasa cuando `h` se acerca a cero.

---

## 4️⃣ ¿Qué es una derivada?

📌 **Definición sencilla**:

> La derivada es la **pendiente de una curva en un punto**

También se puede interpretar como:
- Qué tan rápido cambia una variable
- Cuánto influye `x` sobre `y` en ese punto

Ejemplos:
- Velocidad → derivada de la distancia
- Aceleración → derivada de la velocidad

---

## 5️⃣ Visualizando una derivada con Python

Vamos a graficar:
- Una función curva
- La recta tangente en un punto
- Esa recta representa la derivada



```python
# Explicación visual de derivadas y su relación con regresión lineal

import numpy as np
import matplotlib.pyplot as plt

# Función (una curva)
def f(x):
    return x**2

# Derivada de la función (pendiente en cada punto)
def df(x):
    return 2 * x

# Valores de x
x = np.linspace(-4, 4, 100)
y = f(x)

# Punto donde calculamos la derivada
x0 = 1.5
y0 = f(x0)
pendiente = df(x0)

# Recta tangente (usa la derivada)
y_tangente = pendiente * (x - x0) + y0

# Gráfico
plt.figure()
plt.plot(x, y, label="Función y = x²")
plt.plot(x, y_tangente, linestyle="--", label="Recta tangente (derivada)")
plt.scatter(x0, y0)
plt.text(x0, y0, "  Punto donde calculamos la derivada")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.title("Derivada = pendiente de la recta tangente")

plt.show()
```
