# Función de pérdida (Loss Function)

La función de pérdida mide qué tan mala es la predicción del modelo.

Queremos minimizar esta función durante el entrenamiento.

---

## Regresión

### MSE (Mean Squared Error)
Penaliza fuertemente errores grandes.

$
MSE = \frac{1}{n} \sum (y_{real} - y_{pred})^2
$

### MAE (Mean Absolute Error)
Más robusto ante outliers.

$
MAE = \frac{1}{n} \sum |y_{real} - y_{pred}|
$

---

## Clasificación Binaria

### Binary Cross Entropy

$
BCE = -\frac{1}{n} \sum [ y \log(p) + (1-y)\log(1-p) ]
$

Se usa con activación sigmoid.

---

## Clasificación Multiclase

### Categorical Cross Entropy
Usa etiquetas one-hot.

### Sparse Categorical Cross Entropy
Usa etiquetas enteras (más eficiente).

---

La pérdida es el "objetivo matemático" que queremos minimizar.

# 📌 Regresión — MSE y MAE

```python
import numpy as np

y_true = np.array([3, -0.5, 2, 7])
y_pred = np.array([2.5, 0.0, 2, 8])

mse = np.mean((y_true - y_pred) ** 2)
mae = np.mean(np.abs(y_true - y_pred))

print("MSE:", mse)
print("MAE:", mae)
```

# 📌 Binary Cross Entropy

```python
y_true = np.array([1, 0, 1, 1])
y_pred = np.array([0.9, 0.2, 0.8, 0.7])  # probabilidades

epsilon = 1e-15  # evitar log(0)
y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

bce = -np.mean(
    y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
)

print("Binary Cross Entropy:", bce)
```

# 📌 Multiclase — Cross Entropy (one-hot)

```python
y_true = np.array([
    [1,0,0],
    [0,1,0],
    [0,0,1]
])

y_pred = np.array([
    [0.8,0.1,0.1],
    [0.2,0.7,0.1],
    [0.1,0.2,0.7]
])

epsilon = 1e-15
y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

cce = -np.mean(np.sum(y_true * np.log(y_pred), axis=1))

print("Categorical Cross Entropy:", cce)
```

# 🔹 Con Keras (mucho más simple)

```python
import tensorflow as tf

# --- Regresión (MSE) ---
loss_mse = tf.keras.losses.MeanSquaredError()
y_true_mse = tf.constant([1., 2.], dtype=tf.float32)
y_pred_mse = tf.constant([1.5, 1.8], dtype=tf.float32)
print("MSE:", loss_mse(y_true_mse, y_pred_mse).numpy())

# --- Clasificación binaria (BCE) ---
loss_bce = tf.keras.losses.BinaryCrossentropy(from_logits=False)
y_true_bce = tf.constant([1., 0.], dtype=tf.float32)
y_pred_bce = tf.constant([0.9, 0.2], dtype=tf.float32)
print("BCE:", loss_bce(y_true_bce, y_pred_bce).numpy())

# --- Multiclase sparse (Sparse CCE) ---
loss_sparse = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False)

# y_true: clases enteras (batch,)
y_true_sparse = tf.constant([0, 1], dtype=tf.int32)

# y_pred: probabilidades (batch, num_classes)
y_pred_sparse = tf.constant([
    [0.8, 0.1, 0.1],
    [0.2, 0.7, 0.1]
], dtype=tf.float32)

print("Sparse CCE:", loss_sparse(y_true_sparse, y_pred_sparse).numpy())
```
