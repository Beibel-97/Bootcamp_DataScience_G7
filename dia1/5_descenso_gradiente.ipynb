# Descenso del Gradiente (Gradient Descent)

El **descenso del gradiente** es el algoritmo de optimización utilizado para minimizar una función de pérdida en modelos de machine learning y redes neuronales.

Su objetivo es encontrar los valores de los parámetros (pesos y sesgos) que minimicen el error del modelo.

---

## 🎯 Idea Intuitiva

Imagina que estás en una montaña y quieres bajar hasta el punto más bajo del valle.

- La **función de pérdida** es la montaña.
- El **gradiente** indica la pendiente (dirección de mayor crecimiento).
- Para bajar, debes moverte en la dirección contraria al gradiente.

---

## 📐 Fundamento Matemático

Sea una función de pérdida:

$
L(W)
$

Queremos encontrar el valor de \( W \) que minimiza \( L \).

La actualización de los parámetros es:

$
W := W - \eta \cdot \frac{\partial L}{\partial W}
$

Donde:

- \( W \) = parámetro (peso)
- \( \eta \) = learning rate (tasa de aprendizaje)
- \( \frac{\partial L}{\partial W} \) = gradiente de la pérdida respecto a \( W \)

---

## 📊 ¿Qué es el Gradiente?

El gradiente es la derivada de la función de pérdida respecto a los parámetros.

Indica:
- Magnitud del cambio necesario
- Dirección en la que aumenta más rápido la pérdida

Nosotros nos movemos en la dirección contraria para minimizarla.

---

## ⚙️ Learning Rate (η)

El learning rate controla el tamaño del paso en cada actualización.

- Muy grande → puede divergir
- Muy pequeño → entrenamiento lento
- Correcto → converge eficientemente

---

## 🔁 Proceso Iterativo

El descenso del gradiente funciona de forma iterativa:

1. Inicializar pesos aleatoriamente
2. Calcular predicción (forward pass)
3. Calcular pérdida
4. Calcular gradiente
5. Actualizar pesos
6. Repetir hasta converger

---

## 🧠 Tipos de Descenso del Gradiente

### 1️⃣ Batch Gradient Descent
- Usa todos los datos para calcular el gradiente.
- Estable pero costoso computacionalmente.

### 2️⃣ Stochastic Gradient Descent (SGD)
- Usa un solo ejemplo por actualización.
- Ruidoso pero más rápido.

### 3️⃣ Mini-Batch Gradient Descent
- Usa pequeños lotes de datos.
- Es el más usado en deep learning.

---

## 🚀 Optimizadores Avanzados

El descenso del gradiente clásico puede mejorarse con optimizadores que ajustan dinámicamente el learning rate:

- **Momentum**
- **RMSprop**
- **Adam** (Adaptive Moment Estimation)
- **Adagrad**

Estos métodos aceleran la convergencia y ayudan en superficies complejas.

---

## 📉 Problemas Comunes

- Mínimos locales
- Puntos silla
- Gradientes que explotan
- Gradientes que desaparecen
- Mala elección del learning rate

---

## 📌 En Redes Neuronales

En redes neuronales profundas, el descenso del gradiente trabaja junto con:

- Backpropagation (para calcular gradientes)
- Funciones de activación
- Regularización

TensorFlow automatiza este proceso usando auto-diferenciación.

---

## 🧩 Resumen Conceptual

- La pérdida define qué minimizar.
- El gradiente indica hacia dónde moverse.
- El learning rate define cuánto moverse.
- El descenso del gradiente es el motor que hace que el modelo aprenda.

# EJEMPLO BASICO SIN TENSORFLOW

```python
import numpy as np

# datos simples
X = np.array([[1],[2],[3]])
y = np.array([[2],[4],[6]])

W = np.random.randn(1,1)
b = np.zeros((1,))

lr = 0.01

for epoch in range(1000):
    # Forward
    y_pred = X @ W + b
    loss = np.mean((y - y_pred)**2)

    # Backward (gradientes)
    dW = -2 * X.T @ (y - y_pred) / len(X)
    db = -2 * np.mean(y - y_pred)

    # Update
    W -= lr * dW
    b -= lr * db

print("W:", W, "b:", b)
```

# EJEMPLO BASICO CON TENSORFLOW

```python
import tensorflow as tf
import numpy as np

X = tf.constant([[1.],[2.],[3.]])
y = tf.constant([[2.],[4.],[6.]])

W = tf.Variable(tf.random.normal([1,1]))
b = tf.Variable(tf.zeros([1]))

lr = 0.01

for epoch in range(200):
    with tf.GradientTape() as tape:
        y_pred = tf.matmul(X, W) + b
        loss = tf.reduce_mean((y - y_pred)**2)

    dW, db = tape.gradient(loss, [W, b])

    W.assign_sub(lr * dW)
    b.assign_sub(lr * db)

print("W:", W.numpy(), "b:", b.numpy())
```

# UTILIZANDO KERAS

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=[1])
])

model.compile(optimizer="adam", loss="mse")
model.fit(X.numpy(), y.numpy(), epochs=200, verbose=1)
```
