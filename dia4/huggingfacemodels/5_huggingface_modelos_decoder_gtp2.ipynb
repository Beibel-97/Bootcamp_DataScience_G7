# IMPORTAMOS MODULO PIPELINE

```python
from transformers import pipeline

text_generator = pipeline('text-generation', model='gpt2')
```

# PROBAMOS EL MODLEO GPT2

```python
output = text_generator('el perro ladra y el gato ',max_new_tokens=10)
print(output)
```

```python
output = text_generator('harry potter is', max_new_tokens=50)
print(output[0]['generated_text'])
```
