# Modelos para generación de Imágenes 🖼️

### 📍¿Para qué sirven los modelos de generación de imágenes?

Los modelos de generación de imágenes permiten **crear imágenes nuevas a partir de texto o conceptos**.  
Se utilizan en:
- 🎨 Arte digital y diseño gráfico  
- 👗 Moda y generación de productos  
- 🎮 Videojuegos y animaciones  
- 📚 Educación y contenido visual

Ejemplos populares:
- **Stable Diffusion**  
- **DALLE-3**  
- **Kandinsky**

# IMPORTAMOS LIBRERIAS

```python
from diffusers import StableDiffusionPipeline
import torch
```

# VERIFICARMOS USO DE GPU

```python
# Verificar GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Usando dispositivo:", device)
```

# CARGAMOS EL MODELO

```python
modelo = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    use_safetensors=True
).to(device)
```

# PROBAMOS EL MODELO

```python
prompt = 'a cat with a hat'

with torch.autocast(device):
  simple_img = modelo(prompt).images[0]

simple_img.save('cat_with_hat.png')
display(simple_img)
```
