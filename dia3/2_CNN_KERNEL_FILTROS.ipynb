# 🌟 Redes Neuronales Convolucionales (CNNs)

Las **Redes Neuronales Convolucionales** o **CNNs (Convolutional Neural Networks)** son un tipo de red neuronal especializada en procesar **datos con una estructura de cuadrícula**, como imágenes. Están inspiradas en el **corte visual de los animales**, donde ciertas neuronas se activan al detectar patrones visuales específicos como bordes, formas, etc.

# EJEMPLO DE USO DEL KERNEL(FILTRO) EN IMAGENES

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.ndimage as nd
from skimage import io,color
```

# CARGAMOS FOTO DE EJEMPLO

```python
img = io.imread('/content/neo.jpg')
plt.imshow(img)
```

# CONVERTIR IMAGEN

```python
img_gray = color.rgb2gray(img)
plt.imshow(img_gray)
```

# CREAMOS UN KERNEL (FILTRO)

```python
kernel1 = np.array([[-1,0,-1],
                   [-1,0,-1],
                   [-1,0,-1]])
```

```python
img_new = nd.convolve(img_gray,kernel1)
plt.imshow(img_new)
```

# APLICAMOS OTROS FILTROS

```python
kernel2 = np.array([[ 0, 0, 0],
                   [ 0, 1, 0],
                   [ 0, 0, 0]])
img_new2 = nd.convolve(img_gray,kernel2)
plt.imshow(img_new2)
```

```python
kernel_blur = np.array([[ 0.0625, 0.125, 0.0625],
                   [ 0.125, 0.25, 0.125],
                   [ 0.0625, 0.125, 0.0625]])
img_new3 = nd.convolve(img_gray,kernel_blur)
plt.imshow(img_new3)
```

```python
# Definir diferentes kernels
blur_kernel = np.array([[1, 1, 1],
                        [1, 1, 1],
                        [1, 1, 1]]) / 9  # Promedio (suavizado)

sobel_x_kernel = np.array([[-1, 0, 1],
                            [-2, 0, 2],
                            [-1, 0, 1]])  # Sobel en X (bordes verticales)

laplacian_kernel = np.array([[ 0, -1,  0],
                              [-1,  4, -1],
                              [ 0, -1,  0]])  # Laplaciano (contornos)
```

```python
# Aplicar convoluciones
blurred = nd.convolve(img_gray, blur_kernel)
sobel_x = nd.convolve(img_gray, sobel_x_kernel)
laplacian = nd.convolve(img_gray, laplacian_kernel)

# Mostrar resultados
fig, ax = plt.subplots(1, 4, figsize=(16, 4))
ax[0].imshow(img_gray, cmap="gray")
ax[0].set_title("Original en escala de grises")
ax[0].axis("off")

ax[1].imshow(blurred, cmap="gray")
ax[1].set_title("Suavizado (Blur)")
ax[1].axis("off")

ax[2].imshow(sobel_x, cmap="gray")
ax[2].set_title("Detección de bordes (Sobel X)")
ax[2].axis("off")

ax[3].imshow(laplacian, cmap="gray")
ax[3].set_title("Detección de contornos (Laplaciano)")
ax[3].axis("off")

plt.tight_layout()
plt.show()
```
