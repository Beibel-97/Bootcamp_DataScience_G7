```python
import numpy as np
```

## 🔹 Paso 2: Dataset

```python
texts = [
    "i love this movie",
    "this film is amazing",
    "i hate this movie",
    "this movie is terrible",
    "what a great experience",
    "worst movie ever"
]

labels = [1, 1, 0, 0, 1, 0]
```

## 🔹 Paso 3: Tokenizar manualmente

```python
sentences = [text.split() for text in texts]
```

## 🔹 Paso 4: Entrenar FastText

```python
from gensim.models import FastText

ft_model = FastText(
    sentences,
    vector_size=50,
    window=3,
    min_count=1,
    epochs=50
)
```

## 🔹 Paso 5: Vectorizar frases

```python
def sentence_vector(sentence, model):
    vectors = [model.wv[word] for word in sentence if word in model.wv]
    return np.mean(vectors, axis=0)

X = np.array([sentence_vector(s, ft_model) for s in sentences])
y = np.array(labels)
```

## 🔹 Paso 6: Clasificador

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()
clf.fit(X, y)
```

## 🔹 Paso 7: Predicción

```python
test_sentence = "i love this film".split()
vector = sentence_vector(test_sentence, ft_model)

prediction = clf.predict([vector])
print("Positivo" if prediction[0] == 1 else "Negativo")
```
